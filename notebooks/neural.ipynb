```python
import numpy as np
```

```python
def sigmoid(x):
    return 1 / (1 + np.exp(-x))
```

```python
class SimpleNeuralNetwork:
    def __init__(self):
        # Initialize weights and biases
        self.weights_input_hidden = np.array([[0.15, 0.25, 0.35],  # 2 inputs to 3 hidden neurons
                                            [0.20, 0.30, 0.40]])
        self.bias_hidden = np.array([0.45, 0.50, 0.55])           # Bias for 3 hidden neurons
        self.weights_hidden_output = np.array([0.60, 0.65, 0.70]) # 3 hidden to 1 output
        self.bias_output = np.array([0.75])     
        
        def forward_propagation(self, inputs):
        # Input to hidden layer
        hidden_input = np.dot(inputs, self.weights_input_hidden) + self.bias_hidden
        hidden_output = sigmoid(hidden_input)
        
        # Hidden to output layer
        output_input = np.dot(hidden_output, self.weights_hidden_output) + self.bias_output
        output = sigmoid(output_input)
        
        return output# Bias for output neuron
        
```

```python
if __name__ == "__main__":
    # Create neural network instance
    
    # Example input (2 features)
    input_data = np.array([1.0, 0.5])
    
    # Perform forward propagation
    result = SimpleNeuralNetwork.forward_propagation(input_data)
    print(f"Input: {input_data}")
    print(f"Output: {result}")
```
